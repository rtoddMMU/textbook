DataFrames are a fundamental way of representing data sets. A DataFrame can be viewed in two ways:

- a sequence of named columns that each describe a single attribute of all entries in a data set, or
- a sequence of rows that each contain all information about a single individual in a data set.

## Choosing Sets of Columns
The _double bracket_ noation name_of_dataframe[['column_a','column_b']] creates a new DataFrame consisting of only the specified columns. 
If you assign this _new_ DataFrame to a name, you've just copied your selected data into a new DataFrame. The old one hasn't changed.

```python
cones_noPrice =cones[['Flavor', 'Color']]
cones_noPrice
```

Cones is still unchanged.

```python
cones
```

You can also *drop* columns you don't want. The table above can be created by dropping the `Color` column.

```python
cones.drop('Color', axis = 1 ) # Notce here, it doesn't make sense to drop the column 'Color' unless `axis = 1`!
```

You can name this new table and look at it again by just typing its name.

```python
no_colors = cones.drop('Color',axis=1)

no_colors
```

Like `select`, the `drop` method creates a smaller table and leaves the original table unchanged. In order to explore your data, you can create any number of smaller tables by using choosing or dropping columns. It will do no harm to your original data table.

## Sorting Rows

The `sort_values()` method creates a new DataFrame by arranging the rows of the original table in ascending order of the values in the specified column. 

The Arguments: 
- `by = [list of columns]` this tells you what columns you want to sort by
- `asending = True` list in ascending order. Want descending? Make it `False`. If you leave this off, it will be asceding

Here the `cones` table has been sorted in ascending order of the price of the cones.

```python
cones.sort_values(by = ['Color']) # Notice that when the values are strings it automatically choose alphabetical order!
```

```python
cones.sort_values(by = ['Price'], ascending= True)
```

Like _double backets_ and `drop`, the `sort_values` method leaves the original DataFrame unchanged.

## A small wrinkle: DataFrame vs Series

DataFrames are built to be two diemsional; rows and columns.
- The rows are one dimension (row index)
- The columns are the other dimension (column name)

Even if our DataFrame has only a single column it is built with functionality that allows more than one column

A Series is a one Dimensional object: It can only ever have one column, which means is's basically a labeled array.

When working with a pandas DataFrame, you will sometimes want to access multiple columns or even a single column, but keep the 2D structure. 
Other times you may really only want the single column. 

__How do you get those?__

The choice is in the number of brackets!

- `new_data_frame = old_data_frame[[col_a, col_b, $\ldots$ ]]`: the double brackets must be used because you have more than one column
- `new_data_frame = old_data_frame[[col_a]]`: Here we have only one column, but we're keeping all the structure of a 2D object.
- `new_data_series = old_data_frame[col_a]`: Here we really only care about this array of values so single brackets give us a Series.


```python
cones[['Color']] # Notice that we have rows indexed by integeres starting at 0 and a column named "Color"
```

```python
cones['Color'] # Notice here that we have rows indexed by integers starting at 0 but there's no column name. That information has been disgarded. You can see the name and data-type at the bottom.
```

## Selecting rows that satisfy a condition

Suppose we want all rows that have a specific value in a specific column. You do this all the time when shoping or searching online. Every time you go to a web page and have an option to filter results, you are selecting _rows_ (thing your interested in: socks, hotels, etc.) that satisfy a condition or a collection of conditions. For example, maybe we want to find all the ice cream cones that are pink.

We can easily build a _Series_ of Boolean values that tell us which rows have the "Color" entry as "pink". Once we have this series Pandas allows us to _locate_ the rows whose index is `True` skipping the rows where the index is `False`

```python
cones['Color'] == 'pink' # Series of booleans
```

The `.loc[]` method can be used in many ways. This is one: `df.loc[series of booleans]` filters the DataFrame to only those rows that correspond to `True`.

```python
cones.loc[cones['Color'] == 'pink']
```

Want to do more than one condition? We just need a little __Boolean__ Logic.

Two things at the same time: `(Condition 1) & (Condition 2)` (it's gotta be the amphersand!)
One thing or the other: `(Condition 1) or (Condition 2)`

Let's find all the pink ice cream cones whose price is less than five dolloars

```python
cones.loc[((cones['Color'] == 'pink') & (cones['Price'] < 5.00))]
```

Like all the other table methods in this section, `where` leaves the original table unchanged.

## Example: Salaries in the NBA

"The NBA is the highest paying professional sports league in the world," [reported CNN](http://edition.cnn.com/2015/12/04/sport/gallery/highest-paid-nba-players/) in March 2016. The table `nba` contains the [salaries of all National Basketball Association players](https://www.statcrunch.com/app/index.php?dataid=1843341) in 2015-2016.

Each row represents one player. The columns are:

| **Column Label**   | Description                                         |
|--------------------|-----------------------------------------------------|
| `PLAYER`           | Player's name                                       |
| `POSITION`         | Player's position on team                           |
| `TEAM`             | Team name                                           |
|`SALARY`    | Player's salary in 2015-2016, in millions of dollars|
 
The code for the positions is PG (Point Guard), SG (Shooting Guard), PF (Power Forward), SF (Small Forward), and C (Center). But what follows doesn't involve details about how basketball is played.

The first row shows that Paul Millsap, Power Forward for the Atlanta Hawks, had a salary of almost $\$18.7$ million in 2015-2016.

```python
nba
```

Fans of Stephen Curry can find his row by using `.loc[]`.

```python
nba.loc[nba['PLAYER'] == 'Stephen Curry']
```

We can also create a new table called `warriors` consisting of just the data for the Golden State Warriors.

```python
warriors = nba.loc[nba['TEAM'] == 'Golden State Warriors']
warriors
```

The `nba` table is sorted in alphabetical order of the team names. To see how the players were paid in 2015-2016, it is useful to sort the data by salary. Remember that by default, the sorting is in increasing order.
