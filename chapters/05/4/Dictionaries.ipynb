# Introduction to Python Dictionaries

Dictonaries extend the idea of arrays. Every dictionary has two parts 

- keys
- values

The always come in pairs (key1, value1).

You can almost think about a dictionary as a simple funtion: put in a key, get out a value. Below we'll go through defining simple dictinoaries, accessing their elements, and some basic attributes. 



## 1. Creating a Dictionary

A dictionary is a collection of key-value pairs. Keys must be unique and immutable -that means your keys have to be things that don't change (strings, tuples, floats, etc).
The key structures of a dictinary are:
1. Pyhon recognizes you're building a  dictoinary because you use curly brackets `{}`.
1. Each key value pair is seperated by a colon `:`.
1. Different key-value pairs are seperated by commas `,`.

Check out this dictionary named `student`.

```python
# Creating a dictionary
student = {
    'name': 'Alice',
    'age': 20,
    'major': 'Mathematics'
}
student
```

## 2. Accessing Values by Key

You can access values using square brackets `[]` or the `.get()` method.

```python
# Acces a value like you would in an array
student['name']
```

```python
# Acces a value using the .get() method.
student.get('age')
```

### Why .get() ?
You might wonder why you would use the `.get()` method as opposed to the array-style. There are small differences in the behavior of the two styles that could make a big difference when using them in a large program.

1. If you try and access a key that doesn't exist the array style throws an error
1. If you try and access a key that doesn't exist using `.get()` it returns `None` and just moves on.

Can you think of a scenrio where you might want to use `.get()`?

```python
student.get('gpa') # no problem...nothing happens.
```

```python
student['gpa'] #throws an error. 
```

## 3. Adding and Modifying Entries

Dictionaries are mutable, so you can add or update entries.
- If you want to add a new `key:value` pair, you don't need to redefine the whole dictionary.
- If you want to edit the value associated to a key, you don't need to redefine the whole dictionary.

```python
# Add a new key-value pair
student['gpa'] = 3.8

# Modify existing value
student['age'] = 21
student
```

## 4. Common Dictionary Methods

Here are some useful built-in dictionary methods.

```python
#get all the keys
student.keys()
```

```python
# get all the values
student.values()
```

```python
# get all the key-value pairs
student.items()
```

```python
# pop() - removes a key
student.pop('gpa')
student

# update() - merge another dictionary
student.update({'minor': 'Statistics'})
student
```

## 5. Iterating Through a Dictionary

You can loop through keys, values, or key-value pairs. 

```python
# Loop through keys and values
for key, value in student.items():
    print(key, value)
```
