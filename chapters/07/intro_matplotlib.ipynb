# Introduction to Matplotlib (Minimal Guide)

This short notebook introduces just enough of `matplotlib` to understand a basic bar chart example.

## 1. Importing Matplotlib

We usually import the plotting module as `plt`:

```python
import matplotlib.pyplot as plt
import pandas as pd
```

## 2. Example Data

We will create a small dataset that mimics movie studios and how many movies each has.

```python
data = {
    'Studio': ['Disney', 'Warner Bros', 'Universal', 'Fox', 'Paramount'],
    'count': [15, 12, 10, 8, 5]
}

studio_counts = pd.DataFrame(data)
studio_counts
```

## 3. Sorting the Data

We sort the data so the largest values appear first.

```python
studio_sorted = studio_counts.sort_values(by='count', ascending=False)
studio_sorted
```

## 4. Basic Plot Structure in Matplotlib

Most plots follow this structure:

1. Create a figure
2. Draw a plot (bars, lines, etc.)
3. Add labels and title
4. Show the plot

## 5. Horizontal Bar Chart (`plt.barh`)

- `barh` creates a horizontal bar chart
- First argument → labels (Studios)
- Second argument → values (counts)

```python
plt.figure()

plt.barh(studio_sorted['Studio'], studio_sorted['count'])

plt.xlabel('Count')
plt.ylabel('Studio')
plt.title('Studio Distribution (Top Movies)')

plt.show()
```

## 6. Understanding the Key Lines

- `plt.figure()` → starts a new plot
- `plt.barh(...)` → draws horizontal bars
- `plt.xlabel()` → labels the x-axis
- `plt.ylabel()` → labels the y-axis
- `plt.title()` → adds a title
- `plt.show()` → displays the plot

That’s all you need to understand this example!
