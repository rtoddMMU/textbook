# Minimal Grammar of Graphics with Matplotlib

This notebook introduces a *very small* set of ideas that let you understand most basic plots:

- Data (what you plot)
- Aesthetic mappings (what goes on x, y)
- Geometric objects (bars, lines, points)
- Labels (title and axes)


```python
import matplotlib.pyplot as plt
import pandas as pd
```

## 1. Example Data

We will reuse one dataset and visualize it in different ways.

```python
data = {
    'Year': [2018, 2019, 2020, 2021, 2022],
    'Sales': [100, 120, 90, 150, 170],
    'Profit': [20, 25, 10, 35, 40]
}
df = pd.DataFrame(data)
df
```

## 2. The Basic Plot Template

Almost every matplotlib plot follows this pattern:

1. Create a figure
2. Choose a *geometry* (bar, line, scatter)
3. Map data to axes (x and y)
4. Add labels and title
5. Show the plot


## 3. Scatter Plot (Points)

- Geometry: points
- Use when you want to see relationships between variables

```python
plt.figure()
plt.scatter(df['Sales'], df['Profit'])
plt.xlabel('Sales')
plt.ylabel('Profit')
plt.title('Scatter Plot: Sales vs Profit')
plt.show()
```

## 4. Line Plot (Trends over time)

- Geometry: line
- Use when x has an order (like time)

```python
plt.figure()
plt.plot(df['Year'], df['Sales'])
plt.xlabel('Year')
plt.ylabel('Sales')
plt.title('Line Plot: Sales over Time')
plt.show()
```

## 5. Bar Plot (Comparisons)

- Geometry: bars
- Use to compare categories or discrete values

```python
plt.figure()
plt.bar(df['Year'], df['Sales'])
plt.xlabel('Year')
plt.ylabel('Sales')
plt.title('Bar Plot: Sales by Year')
plt.show()
```

## 6. Horizontal Bar Plot (Your Original Example)

Same idea as bar plots, but flipped for readability.

```python
plt.figure()
plt.barh(df['Year'], df['Sales'])
plt.xlabel('Sales')
plt.ylabel('Year')
plt.title('Horizontal Bar Plot')
plt.show()
```

## 7. Key Idea (Grammar Summary)

Every plot answers:

- What data am I using?
- What goes on the x-axis?
- What goes on the y-axis?
- What geometric shape represents the data?

If you understand these, you can read (and write) most matplotlib code like:

plt.figure()
plt.barh(studio_sorted['Studio'], studio_sorted['count'])
plt.xlabel('Count')
plt.ylabel('Studio')
plt.title('Studio Distribution (Top Movies)')
