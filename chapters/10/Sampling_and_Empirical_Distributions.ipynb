# Sampling and Empirical Distributions

An important part of data science consists of making conclusions based on data in random samples. To correctly interpret results, data scientists must understand exactly what random samples are.

In this chapter we take a closer look at sampling and the properties of large random samples.

Our examples use the `top_movies_2017.csv` dataset.

```python
import pandas as pd
path_data = "../../assets/data/"
import numpy as np
import matplotlib.pyplot as plt
plt.style.use("fivethirtyeight")
%matplotlib inline
```

```python
top = pd.read_csv(path_data + 'top_movies_2017.csv')
top.insert(0, 'Row Index', np.arange(len(top)))
top
```

**Sampling Rows of a DataFrame**

Each row of a data table represents an individual. In `top`, each individual is a movie. Sampling individuals can therefore be accomplished by sampling rows of a DataFrame.

The contents of a row are the values of different variables measured on the same individual. As a result, sampled rows provide samples of the values in each variable.

**Deterministic Samples**

When you explicitly specify which elements to choose, without any randomness involved, you create a *deterministic sample*.

You have done this many times, for example by selecting rows by index:

```python
top.iloc[[3, 18, 100]]
```

You have also used filtering operations:

```python
top[top['Title'].str.contains('Harry Potter', na=False)]
```

While these are samples, they are not random samples. They do not involve chance.

**Probability Samples**

For describing random samples, some terminology is helpful.

A *population* is the set of all elements from which a sample will be drawn.

A *probability sample* is a sample for which it is possible to calculate, before the sample is drawn, the probability that any subset of elements enters the sample.

In a probability sample, not all elements need to have the same probability of selection.

**A Random Sampling Scheme**

Suppose a population consists of three people: A, B, and C. We choose two people according to the following scheme:

- Person A is chosen with probability 1.
- One of B or C is chosen by flipping a fair coin.

This is a probability sample of size 2. The selection probabilities can be calculated exactly and therefore incorporated into any analysis.

**A Systematic Sample**

Imagine all elements of a population listed in sequence. A systematic sample begins with a randomly selected starting position and then selects evenly spaced positions afterward.

Here we choose a random starting row among the first 10 rows and then select every 10th row thereafter.

```python
start = np.random.choice(np.arange(10))
systematic_sample = top.iloc[np.arange(start, len(top), 10)]
systematic_sample
```

Run the cell several times to see how the sample changes.

This systematic sample is a probability sample. Every row has probability $1/10$ of being selected. However, not every subset of rows can occur. Because selected rows are evenly spaced, only certain subsets are possible.

**Random Samples Drawn With or Without Replacement**

In this course, we will mostly use two straightforward sampling methods.

The first is sampling **with replacement**, which is the default behavior of `np.random.choice`.

The second is sampling **without replacement**, often called a *simple random sample*. In this method, selected individuals are not returned to the population before the next draw. This is similar to dealing cards from a deck. To draw a simple random sample using NumPy, specify `replace=False`.

In this chapter we will use simulation to study the behavior of large random samples drawn with and without replacement.

**Convenience Samples**

Drawing a random sample requires careful planning. If you stand on a street corner and record the first ten people who walk by, you have not taken a random sample. Instead, you have taken a *convenience sample*. The probabilities of selection are not known in advance, and the population may not even be clearly defined.
