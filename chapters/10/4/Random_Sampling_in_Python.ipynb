# Random Sampling in Python

This section summarizes several ways to sample at random using Python and pandas, and introduces a method for working directly with categorical distributions.

## Review: Sampling from a Population in a DataFrame

If data for a population are stored in the rows of a pandas DataFrame, you can use the DataFrame method `sample` to randomly select rows. In other words, you can use it to draw a random sample of individuals.

By default, `sample` draws rows uniformly at random without replacement. For sampling with replacement, specify `replace=True`. Sampling with replacement is a natural model for chance experiments such as rolling a die.

```python
import pandas as pd
path_data = "../../../assets/data/"
import matplotlib
matplotlib.use("Agg")
%matplotlib inline
import matplotlib.pyplot as plt
plt.style.use("fivethirtyeight")
import numpy as np
```

```python
faces = np.arange(1, 7)
die = pd.DataFrame({'Face': faces})
die
```

Run the cell below to simulate 7 rolls of a die.

```python
die.sample(n=7, replace=True)
```

Sometimes it is more natural to sample individuals without replacement. This is called a **simple random sample**.

```python
actors = pd.read_csv(path_data + 'actors.csv')
actors
```

```python
# Simple random sample of 5 rows
actors.sample(n=5, replace=False)
```

Because `sample` returns the entire sample, you can use pandas operations to answer many questions about the selected rows. For example, you could count how many sixes appeared in simulated die rolls, compute the average number of movies for sampled actors, or determine whether particular actors appeared in the sample.

## Review: Sampling from a Population Stored as an Array

If a population is represented as a NumPy array, you can use `np.random.choice` to randomly select elements from the array.

By default, `np.random.choice` samples with replacement.

```python
# The faces of a die, as an array
faces
```

```python
# 7 rolls of the die
np.random.choice(faces, 7)
```

The argument `replace=False` allows you to draw a simple random sample without replacement.

```python
actor_names = actors['Actor'].to_numpy()
```

```python
# Simple random sample of 5 actor names
np.random.choice(actor_names, 5, replace=False)
```

Just as with DataFrame sampling, `np.random.choice` returns the entire collection of sampled values. You can use NumPy operations to answer questions about the sample.

## Sampling from a Categorical Distribution

Sometimes we are interested only in category memberships rather than individual observations. For example, we might be interested in whether a coin lands Heads or Tails, or the party affiliations of randomly selected voters.

In these situations, we often care about the proportions of observations in each category. The Data 8 library provides a function called `sample_proportions`. Since we are not using the `datascience` package, we can reproduce the same functionality with NumPy.

The function below takes:

- a sample size
- a population distribution given as proportions that sum to 1

It returns the sample proportions in each category.

```python
def sample_proportions(sample_size, proportions):
    categories = np.arange(len(proportions))
    sample = np.random.choice(categories, size=sample_size, p=proportions)
    counts = np.bincount(sample, minlength=len(proportions))
    return counts / sample_size
```

Suppose a species has red flowers with probability 25%, pink flowers with probability 50%, and white flowers with probability 25%. We can examine the sample proportions among 300 plants.

```python
# Species distribution of flower colors
# Proportions are in the order Red, Pink, White
species_proportions = [0.25, 0.5, 0.25]

sample_size = 300

sample_distribution = sample_proportions(sample_size, species_proportions)
sample_distribution
```

As expected, the sample proportions sum to 1.

```python
sum(sample_distribution)
```

The category order is preserved. Since the categories are ordered as Red, Pink, White, the second entry corresponds to Pink.

```python
# Sample proportion of pink-flowering plants
sample_distribution[1]
```

Using `sample_proportions` and NumPy operations, you can answer questions about the proportions of observations in different categories. However, because the function returns only proportions and not the underlying sample, it cannot answer questions about which specific individuals fell into each category.
