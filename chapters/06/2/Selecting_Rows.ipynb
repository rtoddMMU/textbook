# Selecting Rows with pandas

In this notebook, we learn how to select specific rows from a pandas DataFrame based on position or conditions.

```python
import pandas as pd
import numpy as np
path_data = '../../../assets/data/'
```

## Load the Data

```python
nba = pd.read_csv(path_data + 'nba_salaries.csv')
nba = nba.rename(columns={"'15-'16 SALARY": 'SALARY'})
nba.head()
```

## Selecting Specific Rows by Position
We can select rows using `.iloc[]`, which uses integer positions.

```python
nba.iloc[[0]]
```

```python
nba.iloc[3:6]
```

To get the top 5 highest paid players, we sort first and then take the first 5 rows.

```python
nba.sort_values('SALARY', ascending=False).iloc[0:5]
```

## Selecting Rows by Condition
We can filter rows using boolean conditions.

```python
nba[nba['SALARY'] > 10]
```

Sort those players by salary:

```python
nba[nba['SALARY'] > 10].sort_values('SALARY')
```

Find Stephen Curry's row:

```python
nba.loc[nba['PLAYER'] == 'Stephen Curry']
```

Select all players from a specific team:

```python
nba.loc[nba['TEAM'] == 'Golden State Warriors']
```

## Multiple Conditions
We can combine conditions using `&` (and) and `|` (or).

```python
nba.loc[(nba['POSITION'] == 'PG') & (nba['SALARY'] > 15)]
```

## Range Conditions
We can select rows within a range.

```python
nba.loc[(nba['SALARY'] >= 10) & (nba['SALARY'] < 10.3)]
```

If no rows match, we get an empty DataFrame.

```python
nba.loc[nba['PLAYER'] == 'Barack Obama']
```

## String Filtering
We can filter based on whether text contains a substring.

```python
nba.loc[nba['TEAM'].str.contains('Warriors')]
```

```python
nba.loc[nba['POSITION'].str.contains('G')]
```

## Negation
Use `~` to negate a condition.

```python
nba.loc[~(nba['TEAM'] == 'Cleveland Cavaliers') & (nba['SALARY'] >= 20)]
```

## Summary
- Use `.iloc[]` for position-based selection
- Use `.loc[]` with boolean conditions for filtering
- Combine conditions with `&` and `|`
- Use `.str.contains()` for text matching
- Use `~` for negation
