## Adding Columns

Notice that we added a column by putting a _dictionary_ as the argument for `pd.DataFrame`. That's the most direct way to create a DataFrame, with a dictionary. The keys will be the column names and the values should be np.array's that give the column. 

```python
pd.DataFrame({'Number of petals': np.array([8,34,5])})
```

```python
pd.DataFrame({'Number of petals': np.array([8,34,5]), 'Name': np.array(['lotus','sunflower','rose'])})
```

Below we see another way of adding a column with the method `.assign()`. The stirng the column name is 'Color' and its values are the corresponding array. Why would you do this? Again, it has some special behaviors that are different: `.assign()` always returns an new DataFrame and won't overeight the current one unless you re-assign it. 

```python
flowers = pd.DataFrame({'Number of petals': np.array([8,34,5]), 'Name': np.array(['lotus','sunflower','rose'])})
new_flowers = flowers.assign(Color=np.array(['pink','yellow','red']))
```

```python
new_flowers
```

```python
# We can update the values of the 'Name' column, without changing the original DataFrame.
new_flowers.assign(Name=np.array(['Tom', 'Dick', 'Harry']))
```

```python
new_flowers
```

```python
# If we update the column this way, the original DataFrame is changed
new_flowers['Name'] = np.array(['Tom', 'Dick', 'Harry'])
```

```python
new_flowers
```

### .iloc[]
The .iloc[] accessor in pandas is used for position-based indexing, meaning it selects data based on integer positions rather than labels. You can think of it like Python list slicing applied to rows and columns of a table: the first position is 0, the second is 1, and so on. Inside the brackets, you specify row positions first and column positions second, separated by a comma (for example, df.iloc[0:3, 1]). Slices follow the usual “start:stop” rule, where the start is included and the stop is excluded. This makes .iloc[] especially useful when you want to work with data by its order, regardless of how the rows or columns are labeled.



```python
# all rows, column 4 (fifth column)
minard.iloc[:,4]
```

```python
# entry in row 0 and column 0
minard.iloc[0,4]
```

```python
# Entry in row 5 and column 4
minard.iloc[5,4]
```

## Working with Columns

Here's a sligthly more advanced example. A few things to keep in mind.
- Depending on what you know, you may want to combine different methods of accessing data.
- `.iloc[]` returns a pandas DataFrame if you ask for more than 1 column
- `.iloc[]` returns a series if you ask for exacly one column
- `.iloc[]` can be forced to return a DataFrame of one column. 

Check out the following examples to see the details

```python
 # we asked for two columns, so we must get a DataFrame. Notice the slicing!! start at 0, go upto 4 (but not including) by 2
minard.iloc[:, 0:4:2]
```

```python
# we asked for one column so it gave us a Series
minard.iloc[:, 0] 
```

```python
# We put a second bracket (double bracket!!) around the column index, and instead of a Series we got a DataFrame!
minard.iloc[:, [0]] 
```

```python
#We get thge first row as a series (this is a little weird)
minard.iloc[0, :]
```

```python
# Here we get it as a DataFrame (double-ish brackets again!)
minard.iloc[[0], :]
```

One nice thing about pandas data structures is that they are prepared for you to do arithmeitc on them -they are kind like numpy arrays in that way. 
Look through the next section of code to make sure you see what's going on. 

```python
initial = minard['Survivors'].iloc[0] # this is picking out a number
minard = minard.assign(Percent_Surviving=minard['Survivors']/initial) #this divides all the Survivors column by initial to get a percentage. It's then reassigned!
minard
```

```python
minard.style.format({'Percent_Surviving':'{:.2%}'}) # Here we are updatind the format by rounding to 2 decimals and adding a percent sign (don't worry about the det.s)
```
