# Finding Probabilities

Over the centuries, there has been considerable philosophical debate about what probabilities are. Some people think that probabilities are relative frequencies; others think they are long-run relative frequencies; still others think that probabilities are a subjective measure of uncertainty.

In this course, most probabilities will be interpreted as relative frequencies, though many also have subjective interpretations. Regardless of interpretation, the rules used to calculate and combine probabilities are the same.

By convention, probabilities are numbers between 0 and 1, or equivalently between 0% and 100%. Impossible events have probability 0. Events that are certain have probability 1.

Mathematics is the primary tool for calculating probabilities exactly, though computers can also be very helpful. Simulation often provides excellent approximations.

We will use the notation $P(event)$ to denote the probability that an event occurs.

## When an Event Does Not Happen

If the probability that an event happens is 40%, then the probability that it does not happen is 60%. In general:

$$P(	ext{event does not happen}) = 1 - P(	ext{event happens})$$

## When All Outcomes Are Equally Likely

If all outcomes are equally likely, probabilities can be computed as proportions. For example, when rolling a fair die:

$$
P(\text{even}) = \frac{3}{6}
$$

$$
P(\text{multiple of 3}) = \frac{2}{6}
$$

More generally,

$$
P(\text{event})
=
\frac{\#\{\text{outcomes that satisfy the event}\}}
     {\#\{\text{all outcomes}\}}
$$

## When Two Events Must Both Happen

Suppose a box contains a red, blue, and green ticket. Two tickets are drawn without replacement. The probability of drawing green first and then red is 1/6.

This can also be computed as:

$$
P(\text{green then red})
=
\frac{1}{3} \times \frac{1}{2}
=
\frac{1}{6}
$$

This illustrates the **multiplication rule**:

$$
P(\text{both events happen})
=
P(\text{first event})
\times
P(\text{second event given first})
$$

## When an Event Can Happen in Two Different Ways

Suppose we want the probability that one ticket is green and the other is red, regardless of order. The event can occur as GR or RG.

$$P(	\text{one green and one red}) = P(GR) + P(RG) = \frac{1}{6} + \frac{1}{6} = \frac{2}{6}$$

This illustrates the **addition rule**: when an event can happen in exactly one of several distinct ways, add the probabilities of those ways.

## At Least One Success

A common probability question asks for the chance that an event occurs at least once.

For two coin tosses:

$$P(	\text{at least one head}) = 1 - P(	ext{both tails}) = 1 - \frac{1}{4} = \frac{3}{4}$$

More generally:

$$P(	\text{at least one head in 17 tosses}) = 1 - \left(\frac{1}{2} \right)^{17}$$

For die rolls:

$$P(	\text{at least one 6 in }n	\text{ rolls}) = 1 - \left(\frac{5}{6}
\right)^n$$

The DataFrame below shows these probabilities as the number of rolls increases from 1 to 50.

```python
rolls = np.arange(1, 51, 1)
results = pd.DataFrame({
    "Rolls": rolls,
    "Chance of at least one 6": 1 - (5/6)**rolls
})
results
```

The probability that a 6 appears at least once rises rapidly as the number of rolls increases.

```python
results.plot(x="Rolls", y="Chance of at least one 6", legend=False)
plt.ylabel("Chance of at least one 6")
plt.show()
```

In 50 rolls, you are almost certain to get at least one 6.
