# Randomness

In previous chapters we developed skills needed to describe data. Data scientists must also understand randomness. For example, they may randomly assign individuals to treatment and control groups and then determine whether observed differences are due to chance or to a genuine treatment effect.

In this chapter, we begin our study of randomness. NumPy contains a submodule called `random` with functions for random selection. One of these is `np.random.choice`, which selects an item at random from an array and gives each item an equal chance of being selected.

Thus the following code evaluates to `treatment` with probability 50% and `control` with probability 50%.

```python
import pandas as pd
path_data = "../../../assets/data/"
import numpy as np

import matplotlib
matplotlib.use("Agg")
%matplotlib inline
import matplotlib.pyplot as plt
plt.style.use("fivethirtyeight")
```

```python
two_groups = np.array(['treatment', 'control'])
np.random.choice(two_groups)
```

The code above differs from most previous examples because it does not always return the same value. We can repeat the random selection by providing a second argument indicating how many draws to make.

```python
np.random.choice(two_groups, 10)
```

A fundamental question about random events is whether or not they occur. After an event has occurred, we can often answer a yes-or-no question. In programming, such questions are represented using Boolean values: `True` and `False`.

**Booleans and Comparison**

Boolean values arise most often from comparison operators. For example, `3` is larger than `1 + 1`.

```python
3 > 1 + 1
```

The value `True` indicates that the comparison is valid. Common comparison operators include:

| Comparison | Operator | True Example | False Example |
|------------|----------|--------------|--------------|
| Less than | `<` | `2 < 3` | `2 < 2` |
| Greater than | `>` | `3 > 2` | `3 > 3` |
| Less than or equal | `<=` | `2 <= 2` | `3 <= 2` |
| Greater than or equal | `>=` | `3 >= 3` | `2 >= 3` |
| Equal | `==` | `3 == 3` | `3 == 2` |
| Not equal | `!=` | `3 != 2` | `2 != 2` |

Notice the use of two equal signs (`==`) when checking equality. Python uses a single equal sign (`=`) for assignment, so equality comparisons require a different operator.

```python
5 == 10/2
```

An expression can contain multiple comparisons, and all of them must be true for the overall expression to evaluate to `True`.

```python
1 < 1 + 1 < 3
```

The average of two numbers is always between the smaller and larger number.

```python
x = 12
y = 5
min(x, y) <= (x + y)/2 <= max(x, y)
```

**Comparing Strings**

Strings can also be compared alphabetically.

```python
'Dog' > 'Catastrophe' > 'Cat'
```

Returning to random selection, we can check whether a randomly selected individual was assigned to the treatment group.

```python
np.random.choice(two_groups) == 'treatment'
```

Because the selection is random, the result of the comparison can vary each time the cell is executed.

**Comparing an Array and a Value**

When a NumPy array is compared with a single value, each element is compared separately and the result is an array of Boolean values.

```python
tosses = np.array(['Tails', 'Heads', 'Tails', 'Heads', 'Heads'])
tosses == 'Heads'
```

The NumPy function `count_nonzero` counts the number of `True` values in a Boolean array.

```python
np.count_nonzero(tosses == 'Heads')
```
