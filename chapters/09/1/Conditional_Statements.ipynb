# Conditional Statements

In many situations, actions and results depend on a specific set of conditions being satisfied. For example, individuals in randomized controlled trials receive the treatment if they have been assigned to the treatment group. A gambler makes money if she wins her bet.

In this section we will learn how to describe such situations using Python code. A *conditional statement* is a multi-line statement that allows Python to choose among different alternatives based on the truth value of an expression. While conditional statements can appear anywhere, they appear most often within the body of a function in order to express alternative behavior depending on argument values.

A conditional statement always begins with an `if` header, which is a single line followed by an indented body. The body is executed only if the expression following `if` evaluates to `True`. If it evaluates to `False`, the body is skipped.

Let us start by defining a function that returns the sign of a number.

````python
import pandas as pd
path_data = "../../../assets/data/"

import matplotlib
matplotlib.use("Agg")
%matplotlib inline
import matplotlib.pyplot as plt
plt.style.use("fivethirtyeight")
import numpy as np
````

````python
def sign(x):
    if x > 0:
        return "Positive"
````

````python
sign(3)
````

This function returns the correct sign if the input is a positive number. But if the input is not a positive number, the `if` expression evaluates to `False`, the `return` statement is skipped, and the function returns no value.

````python
sign(-3)
````

So let us refine our function to return `Negative` if the input is a negative number. We can do this by adding an `elif` clause. The keyword `elif` is short for "else if."

````python
def sign(x):
    if x > 0:
        return "Positive"
    elif x < 0:
        return "Negative"
````

Now `sign` returns the correct answer when the input is -3:

````python
sign(-3)
````

What if the input is 0? To deal with this case, we can add another `elif` clause:

````python
def sign(x):
    if x > 0:
        return "Positive"
    elif x < 0:
        return "Negative"
    elif x == 0:
        return "Neither positive nor negative"
````

````python
sign(0)
````

Equivalently, we can replace the final `elif` clause with an `else` clause. The `else` body will execute only if all previous conditions are false.

````python
def sign(x):
    if x > 0:
        return "Positive"
    elif x < 0:
        return "Negative"
    else:
        return "Neither positive nor negative"
````

````python
sign(0)
````

## The General Form

A conditional statement can contain multiple clauses, but only one body will ever be executed.

```python
if <if expression>:
    <if body>
elif <elif expression 0>:
    <elif body 0>
elif <elif expression 1>:
    <elif body 1>
...
else:
    <else body>
```

There is always exactly one `if` clause, but there can be any number of `elif` clauses. Python evaluates conditions in order until it finds one that is `True`, then executes the corresponding body. The `else` clause is optional and runs only if none of the previous conditions are true.

## Example: Betting on a Die

Suppose I bet on a roll of a fair die. The rules of the game are:

- If the die shows 1 or 2 spots, I lose a dollar.
- If the die shows 3 or 4 spots, I neither lose nor gain money.
- If the die shows 5 or 6 spots, I gain a dollar.

We will now use conditional statements to define a function `one_bet` that takes the number of spots and returns my net gain.

````python
def one_bet(x):
    """Returns my net gain if the die shows x spots"""
    if x <= 2:
        return -1
    elif x <= 4:
        return 0
    elif x <= 6:
        return 1
````

Let's check that the function behaves correctly for each possible outcome.

````python
one_bet(1), one_bet(2), one_bet(3), one_bet(4), one_bet(5), one_bet(6)
````

As a review of how conditional statements work, consider what happens when the input is 3:

- The condition `3 <= 2` is `False`, so the first body is skipped.
- The condition `3 <= 4` is `True`, so Python executes that body and returns 0.
- No later conditions are evaluated.

If the input is greater than 6, all conditions evaluate to `False`, so the function returns no value.

````python
one_bet(17)
````

To play the game based on one roll of a die, generate a random integer from 1 through 6 using NumPy and pass the result to `one_bet`. Run the cell several times to see different outcomes.

````python
one_bet(np.random.choice(np.arange(1, 7)))
````
